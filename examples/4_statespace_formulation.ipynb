```python
import numpy as np

from pybatteryid.utilities import load_model_from_file

from pybatteryse.statespace import StateSpace
```

### Formulating state-space representations of the LPV battery models

We load an example model to understand the state-space representations that can be formulated using the LPV models identified using [`PyBatteryID`](https://github.com/USERNAME/PyBatteryID).

```python
model = load_model_from_file('data/nmc_with_temperature/model_n,l=3,3.npy')
```

#### Case 1. When the state vector contains `s` but not overpotentials.

It is certainly possible to formulate a state-space representation without converting the identified models to state-space representation. Essentially, we can define the state using SOC and its past values, and simply using input--output equation to calculate the voltage prediction. Note that input--output equation does require past voltage values to calculate the predicted output.

We can load the state-space representation using the `StateSpace` class as follows.

```python
state_components = ['s', 'overpotentials']
ss_representation = StateSpace(model=model, state_components=state_components)
```

And we can print the state-space structure as follows.

```python
ss_representation.print_equations()
```
